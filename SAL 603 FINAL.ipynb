```python
# ============================================================
# SECTION 3: DATA TRANSFORMATIONS
# ============================================================

# Map 5 positions to 3 broader position groups
pos_map = {'PG': 'Guard', 'SG': 'Guard', 'SF': 'Forward', 'PF': 'Forward', 'C': 'Center'}
df['Pos_Group'] = df['Position'].map(pos_map)

# Create Net Contribution feature (positive stats minus turnovers)
df['Net_Contrib'] = df['PTS'] + df['REB'] + df['AST'] + df['STL'] + df['BLK'] - df['TOV']

print("── Dataset Overview ──────────────────────────────────")
print(f"Shape         : {df.shape[0]} players × {df.shape[1]} columns")
print(f"Position groups:\n{df['Pos_Group'].value_counts().to_string()}")
print()
print("── Descriptive Statistics ────────────────────────────")
print(df[['PTS', 'REB', 'AST', 'TOV', 'FG_PCT', 'PER']].describe().round(2).to_string())
print()
```

```python
# ============================================================
# SECTION 4: RQ1 — CORRELATION ANALYSIS
# Visualization 1: Correlation Heatmap
# ============================================================

print("── RQ1: Correlations with PER ───────────────────────")
features = ['PTS', 'REB', 'AST', 'STL', 'BLK', 'TOV', 'FG_PCT', 'FT_PCT', 'PER']
corr_matrix = df[features].corr()

per_corr = corr_matrix['PER'].drop('PER').sort_values(ascending=False)
for stat, val in per_corr.items():
    bar = '█' * int(abs(val) * 20)
    sign = '+' if val > 0 else '-'
    print(f"  {stat:8s}: {sign}{abs(val):.3f}  {bar}")
print()

# Plot
fig, ax = plt.subplots(figsize=(10, 8))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(
    corr_matrix, mask=mask, annot=True, fmt='.2f',
    cmap='RdYlGn', center=0, vmin=-1, vmax=1,
    square=True, linewidths=0.5, ax=ax,
    annot_kws={'size': 10}
)
ax.set_title(
    'Correlation Matrix: NBA Box Score Stats vs PER\n(2023-24 Season)',
    fontsize=14, fontweight='bold', pad=15
)
plt.tight_layout()
plt.savefig('viz1_correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print("Visualization 1 saved: viz1_correlation_heatmap.png\n")
```

```python
# ============================================================
# SECTION 5: RQ2 — POSITION GROUP ANALYSIS
# Visualization 2: PER Distribution by Position + Scatter
# ============================================================

print("── RQ2: PER by Position Group ───────────────────────")
pos_stats = df.groupby('Pos_Group')['PER'].agg(['mean', 'median', 'std', 'min', 'max']).round(2)
print(pos_stats.to_string())
print()

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

pos_order = ['Guard', 'Forward', 'Center']
colors    = {'Guard': '#E63946', 'Forward': '#2A9D8F', 'Center': '#457B9D'}

# Left — violin + scatter overlay
for i, pos in enumerate(pos_order):
    subset = df[df['Pos_Group'] == pos]['PER']
    parts  = axes[0].violinplot(subset, positions=[i], widths=0.6,
                                showmeans=True, showmedians=True)
    for pc in parts['bodies']:
        pc.set_facecolor(colors[pos])
        pc.set_alpha(0.7)
    axes[0].scatter([i] * len(subset), subset,
                    color=colors[pos], alpha=0.85, s=55, zorder=3)

axes[0].set_xticks([0, 1, 2])
axes[0].set_xticklabels(pos_order, fontsize=11)
axes[0].set_ylabel('Player Efficiency Rating (PER)', fontsize=11)
axes[0].set_title('PER Distribution by Position', fontsize=13, fontweight='bold')
axes[0].axhline(15, color='gray', linestyle='--', alpha=0.6, label='League Avg (15.0)')
axes[0].legend(fontsize=9)

# Right — PTS vs PER scatter
for pos, group in df.groupby('Pos_Group'):
    axes[1].scatter(group['PTS'], group['PER'],
                    label=pos, color=colors[pos],
                    s=80, alpha=0.85, edgecolors='white', linewidth=0.5)

# Trend line
z       = np.polyfit(df['PTS'], df['PER'], 1)
x_line  = np.linspace(df['PTS'].min(), df['PTS'].max(), 100)
axes[1].plot(x_line, np.poly1d(z)(x_line), 'k--', alpha=0.5, linewidth=1.5, label='Trend')

# Label notable players
for _, row in df[df['PER'] > 27].iterrows():
    axes[1].annotate(row['Player'], (row['PTS'], row['PER']),
                     textcoords='offset points', xytext=(5, 4),
                     fontsize=8, style='italic')

axes[1].set_xlabel('Points Per Game', fontsize=11)
axes[1].set_ylabel('Player Efficiency Rating (PER)', fontsize=11)
axes[1].set_title('Points Per Game vs PER by Position', fontsize=13, fontweight='bold')
axes[1].legend(fontsize=9)

plt.suptitle('Research Q2: Do Position Groups Differ in PER Drivers?',
             fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('viz2_position_analysis.png', dpi=150, bbox_inches='tight')
plt.show()
print("Visualization 2 saved: viz2_position_analysis.png\n")
```

```python
# ============================================================
# SECTION 6: RQ3 — REGRESSION MODEL
# ============================================================

print("── RQ3: Multiple Linear Regression ─────────────────")

feature_cols = ['PTS', 'REB', 'AST', 'STL', 'BLK', 'TOV', 'FG_PCT', 'FT_PCT']
X = df[feature_cols]
y = df['PER']

# Standardize features so coefficients are comparable
scaler   = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 70/30 train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.30, random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

r2   = r2_score(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print(f"  R² Score            : {r2:.4f}")
print(f"  RMSE (PER points)   : {rmse:.4f}")
print(f"  Variance explained  : {r2 * 100:.1f}%")
print()

# Coefficient table
coef_df = pd.DataFrame({
    'Statistic':   feature_cols,
    'Coefficient': model.coef_
}).sort_values('Coefficient', ascending=False).reset_index(drop=True)

print("  Standardized Regression Coefficients:")
print(coef_df.to_string(index=False))
print()
```

```python
# ============================================================
# SECTION 7: RQ3 — VISUALIZATION
# Visualization 3: Coefficients + Actual vs Predicted
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Left — coefficient bar chart
bar_colors = ['#2A9D8F' if c > 0 else '#E63946' for c in coef_df['Coefficient']]
bars = axes[0].barh(coef_df['Statistic'], coef_df['Coefficient'],
                    color=bar_colors, edgecolor='white', linewidth=0.5)
axes[0].axvline(0, color='black', linewidth=1)
axes[0].set_xlabel('Standardized Coefficient', fontsize=11)
axes[0].set_title('Regression Coefficients\n(Impact on PER per 1 SD change)',
                  fontsize=12, fontweight='bold')
axes[0].legend(handles=[
    mpatches.Patch(color='#2A9D8F', label='Positive impact'),
    mpatches.Patch(color='#E63946', label='Negative impact')
], fontsize=9)
for bar, val in zip(bars, coef_df['Coefficient']):
    axes[0].text(
        val + (0.05 if val >= 0 else -0.05),
        bar.get_y() + bar.get_height() / 2,
        f'{val:.2f}', va='center',
        ha='left' if val >= 0 else 'right', fontsize=9
    )

# Right — actual vs predicted
y_all_pred = model.predict(X_scaled)
scatter = axes[1].scatter(y, y_all_pred, c=df['PER'], cmap='RdYlGn',
                          s=80, edgecolors='white', linewidth=0.5, zorder=3)
fig.colorbar(scatter, ax=axes[1], label='Actual PER')
lims = [min(y.min(), y_all_pred.min()) - 1, max(y.max(), y_all_pred.max()) + 1]
axes[1].plot(lims, lims, 'k--', alpha=0.6, linewidth=1.5, label='Perfect Prediction')
axes[1].set_xlim(lims)
axes[1].set_ylim(lims)
axes[1].set_xlabel('Actual PER', fontsize=11)
axes[1].set_ylabel('Predicted PER', fontsize=11)
axes[1].set_title(
    f'Actual vs Predicted PER\n(R² = {r2_score(y, y_all_pred):.3f})',
    fontsize=12, fontweight='bold'
)
axes[1].legend(fontsize=9)
for i, row in df.iterrows():
    if row['PER'] > 27 or row['PER'] < 10:
        axes[1].annotate(row['Player'], (row['PER'], y_all_pred[i]),
                         textcoords='offset points', xytext=(5, 3),
                         fontsize=7.5, style='italic')

plt.suptitle('Research Q3: Multiple Regression Model for PER Prediction',
             fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('viz3_regression_results.png', dpi=150, bbox_inches='tight')
plt.show()
print("Visualization 3 saved: viz3_regression_results.png\n")
```
