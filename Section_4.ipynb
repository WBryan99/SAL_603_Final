```python
# Cell 1: Setup and imports
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

import pandas as pd
```

```python
# Cell 2: Choose a league, season, and 4 teams within the league
# League: NBA
# Season: 2023-24 (or the season from your CSV files)
# Teams: Raptors, 76ers, Nets, Celtics,
# Note: The roster tables have already been collected from Sports Reference
# and are provided as CSV files (Raptors.csv, 76ers.csv, Nets.csv, Celtics.csv, Knicks.csv)

print("Using 4 NBA teams from the Atlantic Division:")
print("- Toronto Raptors")
print("- Philadelphia 76ers")
print("- Brooklyn Nets")
print("- Boston Celtics")
```

```python
# Cell 3: Read all 4 CSVs with the index being the player name
# Note: Using 4 teams as specified in the assignment (excluding Knicks for now)
raptors_df = pd.read_csv('Raptors.csv', index_col='Player')
sixers_df = pd.read_csv('76ers.csv', index_col='Player')
nets_df = pd.read_csv('Nets.csv', index_col='Player')
celtics_df = pd.read_csv('Celtics.csv', index_col='Player')

print("Raptors roster:")
print(raptors_df)
print("\n" + "="*80 + "\n")

print("76ers roster:")
print(sixers_df)
print("\n" + "="*80 + "\n")

print("Nets roster:")
print(nets_df)
print("\n" + "="*80 + "\n")

print("Celtics roster:")
print(celtics_df)
print("\n" + "="*80 + "\n")
```

```python
# Cell 4: Concatenate the 4 tables into 1 all-inclusive table
# Concatenate all four dataframes into one combined table
all_rosters = pd.concat([raptors_df, sixers_df, nets_df, celtics_df])

print("CONCATENATED TABLE:")
print("="*80)
print(f"Total number of rows (players): {len(all_rosters)}")
print(f"Number of columns: {len(all_rosters.columns)}")
print(f"\nColumn names: {list(all_rosters.columns)}")
print("\n" + "="*80)
print("\nFirst 10 rows of combined roster:")
print(all_rosters.head(10))
print("\n" + "="*80)
print("\nLast 10 rows of combined roster:")
print(all_rosters.tail(10))
print("\n" + "="*80)
```

```python
# Cell 5: Write the concatenated table to CSV (Note: the index SHOULD be included)
# Write the combined roster to a CSV file with the index (player names) included
output_filename = 'all_teams_roster.csv'
all_rosters.to_csv(output_filename, index=True)
print(f"✓ Combined roster successfully saved to '{output_filename}'")
print(f"✓ Index (Player names) included: Yes")
print(f"✓ Total rows written: {len(all_rosters)}")
print(f"✓ Total columns written: {len(all_rosters.columns) + 1}")  # +1 for index column

# Verify the file was written correctly
verify_df = pd.read_csv(output_filename)
print(f"\nVerification - File '{output_filename}' contains {len(verify_df)} rows")
```

```python
# Cell 6: Final summary and verification
# Display summary statistics of the final dataset
print("="*80)
print("FINAL ASSIGNMENT SUMMARY")
print("="*80)

print("\n📊 DATASET OVERVIEW:")
print(f"   • Total players across all teams: {len(all_rosters)}")
print(f"   • Number of teams: 4")
print(f"   • Teams included: Raptors, 76ers, Nets, Celtics")
print(f"   • Output file: {output_filename}")

print("\n📋 PLAYERS PER TEAM:")
if 'Tm' in all_rosters.columns:
    print(all_rosters['Tm'].value_counts())
else:
    print("   Team information available in original CSV files")

```
